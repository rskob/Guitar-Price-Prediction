```python
import pandas as pd
import sqlite3
import json
```

```python
conn = sqlite3.connect("../data.db")
def get_df(conn, shop_name):   
    with conn:
        cur = conn.execute(f"SELECT json_data from raw_data WHERE shop_name = '{shop_name}'")
        json_list = [json.loads(row[0]) for row in cur.fetchall()]

    return pd.DataFrame(json_list)

df_skif = get_df(conn, "SKIFMUSIC")
df_skif.columns
```

```python
new_names = {column: column.replace(":", "") for column in df_skif.columns.to_list()}
df_skif.rename(columns=new_names, inplace=True)
df_skif = df_skif.drop(columns=["Код товара", "Тип", "Состояние", "Уровень инструмента", "Артикул", "Цвет корпуса", "Название", "В комплекте"])
df_skif.columns
```
